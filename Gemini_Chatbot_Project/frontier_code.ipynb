```python
from transformers import pipeline

# Load model
pipe = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    device_map="auto"
)

# Memory storage
oss_memory = []

# Safety words
blocked_words = ["bomb", "hack", "kill", "attack", "weapon"]

# Assistant function
def oss_chat(user_input):

    # Safety check
    if any(word in user_input.lower() for word in blocked_words):
        return "I cannot assist with harmful or unsafe requests."

    # Add user message to memory
    oss_memory.append(f"User: {user_input}")

    # Keep only last 6 messages to avoid long prompts
    conversation = "\n".join(oss_memory[-6:])

    # System prompt
    prompt = f"""
You are a helpful, polite AI assistant.
Answer clearly and briefly.

{conversation}

Assistant:
"""

    # Generate response
    response = pipe(
        prompt,
        max_new_tokens=80,
        do_sample=False,
        truncation=True,
        temperature=0.7
    )

    # Extract assistant reply
    generated_text = response[0]["generated_text"]

    reply = generated_text.split("Assistant:")[-1].strip()

    # Save assistant reply
    oss_memory.append(f"Assistant: {reply}")

    return reply

print("OSS Assistant Ready")
```

```python
print(oss_chat("Hello"))

print(oss_chat("My name is Vishnu"))

print(oss_chat("What is my name?"))

print(oss_chat("Tell me a joke"))
```
